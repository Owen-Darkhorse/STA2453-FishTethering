```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('C:\\Users\\86139\\Desktop\\FishTetherExperiment\\ProcessedData\\acousticDataFrame.csv')
fishList = ["LT001", "BUR001", "LWF001", "SMB001"]
freqList = ["F86.5", "F173"]
```

```python
ltData = df.loc[df['fishNum'] == "LT001", freqList]
burData = df.loc[df['fishNum'] == "BUR001", freqList]
lwfData = df.loc[df['fishNum'] == "LWF001", freqList]
smbData = df.loc[df['fishNum'] == "SMB001", freqList]
```

```python
## BUR has the shortest length, test trying FFT on it
n = len(burData["F86.5"])
t = np.arange(0, n, 1)

burFFT = np.fft.fft(burData["F86.5"])
psdBUR = abs(burFFT) / n
freq = np.arange(n)
# L = np.arange(1, np.floor(n/2), dtype='int')

fig, axs = plt.subplots(2, 1)
plt.sca(axs[0])
plt.plot(t, burData["F86.5"], color = "c", label = "F86.5")
plt.legend()

plt.sca(axs[1])
plt.plot(freq, psdBUR, color = "c", label = "F86.5")
plt.legend()

plt.show()
```
